# Activation Functions

## Introduction

Activation functions are the functions that adds non-linearity to the output of layer. Previously we learned a neural network is combination of several linear functions / layers. Each layer's output is linear. However, sometimes the dataset can be too complex to learn with linear layers only. So, by adding non-linearity, the network can learn complex data.

Some common activation functions are as follow:
- Rectified Linear Unit (ReLU): This is the most common activation function used in dense neural network. It clips negative outputs to zero
- Sigmoid: This is genrally used in binary classification task. It uses sigmoid function to return probabilities for classes (0, 1).
- Softmax: This activation function is used for multiclass classification task. Similar to sigmoid, it also returns probabilities for classes (1, n)
- Heavyside
- Tanh
